```python
import pandas as pd
from collections import Counter
```

```python
criteria_df = pd.read_csv(r"${DEMENTIA_DATA_ROOT}\dementia_detection\medical_data\study_criteria_table.csv")
```

```python
criteria_df
```

```python
criteria_df.columns
```

```python
EDW_Enc_df = pd.read_csv(r"${DEMENTIA_DATA_ROOT}\dementia_detection\medical_data\EDW_EncounterDiagnosis.csv",encoding='utf-8')
RPDR_Enc_df = pd.read_csv(r"${DEMENTIA_DATA_ROOT}\dementia_detection\medical_data\RPDR_Dia_All.csv",encoding='utf-8')
```

```python
EDW_Enc_df
```

```python
RPDR_Enc_df = RPDR_Enc_df.dropna(subset=['Diagnosis_Name'])
```

```python
RPDR_Enc_df
```

# Dementia ICDs

```python
dementia_PatientID = list(criteria_df[criteria_df['Dementia_Enc'] ==1]['PatientID'])
dementia_EMPI = list(criteria_df[criteria_df['Dementia_Enc'] ==1]['EMPI'])
```

```python
len(dementia_PatientID)
```

```python
Dementia_EDW_Enc_df = EDW_Enc_df[EDW_Enc_df.PatientID.isin(dementia_PatientID)]
```

```python
Dementia_EDW_Enc_df = Dementia_EDW_Enc_df.drop_duplicates(subset=['PatientID','DiagnosisNM'])
```

```python
Dementia_EDW_Enc_df
```

```python
Dementia_RPDR_Enc_df = RPDR_Enc_df[RPDR_Enc_df.EMPI.isin(dementia_EMPI)]
```

```python
Dementia_RPDR_Enc_df
```

```python
Dementia_Counter = Counter(Dementia_EDW_Enc_df['DiagnosisNM'])
```

```python
Dementia_Counter_df = pd.DataFrame.from_dict(Dementia_Counter , orient='index').reset_index()
```

```python
Dementia_Counter_df.columns = ['DiagnosisNM','Count']
```

```python
Dementia_Counter_df = Dementia_Counter_df.sort_values(by=['Count'],ascending=False)
```

```python
Dementia_Counter_df[Dementia_Counter_df['DiagnosisNM'].str.contains('Dementia')]
```

```python
Dementia_Counter_df[Dementia_Counter_df['Count'] > 10][:50]
```

```python
Dementia_Counter_df[Dementia_Counter_df['Count'] > 10].to_csv('..\medical_data\Dementia_EDW_ICD.csv',index=False)
```

```python
Dementia_Counter = Counter(Dementia_RPDR_Enc_df['Diagnosis_Name'])
```

```python
Dementia_Counter_df = pd.DataFrame.from_dict(Dementia_Counter , orient='index').reset_index()
```

```python
Dementia_Counter_df.columns = ['Diagnosis_Name','Count']
```

```python
Dementia_Counter_df = Dementia_Counter_df.sort_values(by=['Count'],ascending=False)
```

```python
Dementia_Counter_df[Dementia_Counter_df['Diagnosis_Name'].str.contains('Dementia')]
```

```python
Dementia_Counter_df[Dementia_Counter_df['Count'] > 10].to_csv('..\medical_data\Dementia_RPDR_ICD.csv',index=False)
```
