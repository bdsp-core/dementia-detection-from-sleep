```python
import warnings
warnings.filterwarnings("ignore")
#define the lasso model
def myfit(X, y, sids, random_state=2019):
    coefs = []
    X = (X-X.mean(axis=0)) / X.std(axis=0)

    # do the actual kfold lasso cv and calculate the mean alpha. 
    from sklearn.model_selection import KFold
    k_fold = KFold(10, shuffle=True)
    mses = []
    alphas = []
    ytes = []
    yptes = []
    l1_ratios = []
    sids_tes = []
    #split data in train set and test set
    for k, (train, test) in enumerate(k_fold.split(X, y)):
        X_train = X[train]
        X_test = X[test] 
        y_train = y[train]
        yte = y[test]
        sids_te = sids[test]
        print('X_train shape', X_train.shape)
        #check for correlation pvals<0.1
        pvals = []
        for i in range(X_train.shape[1]):
            # check if the i-th feature is binary
            #X[:,i]

            pval = pearsonr(X_train[:,i], y_train)[1]
            pvals.append(pval)

        pvals = np.array(pvals)
        select_ids = np.sort(np.argsort(pvals)[:60])#np.where(pvals<0.1)[0]
        print('fold', k+1)
        print('before univariate selection', X_train.shape[1])
        X_train = X_train[:,select_ids]
        X_test = X_test[:,select_ids]
        #print(select_ids)
        print('after univariate selection', X_train.shape[1])
        
        # do lasso
        lasso_cv = LassoCV(cv=10, max_iter=2000, tol=0.001, random_state=2019, normalize=False)
        lasso_cv.fit(X_train, y_train)
        ypte = lasso_cv.predict(X_test)
        score = np.mean((yte - ypte)**2)
        ytes.extend(yte)
        yptes.extend(ypte)
        sids_tes.extend(sids_te)
        mses.append(score)
        alphas.append(lasso_cv.alpha_)
        
    mean_mse = np.median(mses)
    mean_alpha = np.median(alphas)

    # now we do one last lasso, where we use the mean alpha as alpha. 
    pvals = []
    for i in range(X.shape[1]):
        # check if the i-th feature is binary
        #X[:,i]

        pval = pearsonr(X[:,i], y)[1]
        pvals.append(pval)
            
    pvals = np.array(pvals)
    select_ids = np.sort(np.argsort(pvals)[:60])
    Xnames2 = Xnames[select_ids]
    X = X[:, select_ids]
        
    lasso = Lasso(alpha=mean_alpha, max_iter=100000, normalize=False)
    lasso.fit(X, y)
    #yp_final = lasso.predict(X)
    print(mses)
    print(alphas)
    print(lasso.coef_)
    return lasso.coef_, mean_mse, mean_alpha,  yptes, ytes, sids_tes, Xnames2
```

```python
# do bootstrap to get the confidence interval around the coefs 
np.random.seed(2019)
n_iterations = 0#2
n_size = int(len(Brain_age) * 0.50)
# run bootstrap
stats = []
N = len(X) ######
coefs = []
mses  = []
alphas = []

for i in range(n_iterations+1):
    # generate bootstrapping ids
    ids = np.random.choice(N, N, replace=True)
    
    #to check how far your model is, you can let it print each cycle
    #if i%1==0:
    print(i)
    
    if i==0:
        Xbt = X
        ybt = y
        sidsbt = sids
    else:
        # use the bootstrapping ids to index X and y
        Xbt = X[ids]
        ybt = y[ids]
        sidsbt = sids[ids]
    
    # fit the model using Xbt and ybt
    coef, mse, alpha, yptes_, ytes_, sids_tes_, Xnames2_ = myfit(Xbt, ybt, sidsbt)
    if i==0:
        yptes = yptes_
        ytes = ytes_
        sids_tes = sids_tes_
        Xnames2 = Xnames2_
        print('Xnames2.shape = ', Xnames2.shape)
    
    # append to coefs
    coefs.append(coef)
    mses.append(mse)
    alphas.append(alpha)
    
# compute the upper and lower bound for the confidence interval of each coef
alpha = 0.95
p = ((1.0-alpha)/2.0) * 100
lower = np.percentile(coefs, p, axis=0)
p = (alpha+((1.0-alpha)/2.0)) * 100
upper = np.percentile(coefs, p, axis=0)

print('coefs[0].shape', coefs[0].shape)
big_ids = np.where(np.abs(coefs[0])>1e-6)[0]
output = (np.c_[np.array(Xnames2)[big_ids], coefs[0][big_ids]])#, lower[big_ids], upper[big_ids]])

# convert from np.array to pd.dataframe
dataset = pd.DataFrame({'Variable': output[:, 0], 'coefs': output[:, 1]})#, 'lower': output[:, 2], 'upper': output[:, 3]})
print(dataset)
dataset.to_csv("${DEMENTIA_DATA_ROOT}/brain_age_computer_files/coefs_NIH_8.csv",index=False)

# calculate h-value (haoqi value, optional. Probably not the best way to calculate some sort of p-value, but this is what we came up with).
#for i in big_ids:
 #   hval = np.sum(coefs[1:,i]==0)/1000
  #  print(hval)

# save yptes as BHI
```
