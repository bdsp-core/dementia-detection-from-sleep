```python
def benjamini_hochberg(raw_p_values, alpha):
    # compare each member of sorted p values with alpha*i/m, where n is total number of tests, i is p values index
    p_vals_sorted = sorted(raw_p_values)
    m = len(raw_p_values)
    comparison_vals = [alpha*(ii+1)/m for ii in range(len(raw_p_values))]
    boolean_is_significant = [p_vals_sorted[ii] <= comparison_vals[ii] for ii in range(len(raw_p_values))]
    # get highest p value that is small enough to reject
    last_index = max(  [ii for ii, x in enumerate(boolean_is_significant) if x == True] )
    top_p_val = p_vals_sorted[last_index]
    # any p value smaller than top_p_val indicates rejected null
    p_vals_below = list(set(p_vals_sorted[:last_index + 1]))
    return p_vals_below, top_p_val
```
