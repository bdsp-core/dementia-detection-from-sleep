```python
### Luna features
luna_df = pd.read_csv('../eeg_data/combined_luna_output.csv')
```

```python
#eeg_df = pd.read_csv(r'..\eeg_data\eeg_list.csv')
eeg_df = pd.read_csv(r'../eeg_data/eeg_list.csv')
```

```python
ID_df = eeg_df[['FolderName','SignalName']]
```

```python
ID_df = ID_df.dropna()
```

```python
ID_df['ID'] = [row['SignalName'].lower()[:-4] for index,row in ID_df.iterrows()]
```

```python
ID_df['ID'] = [row['ID'].replace(' ','_') for index,row in ID_df.iterrows()]
```

```python
ID_df
```

```python
luna_df = luna_df.dropna(subset=['AMP'])
```

```python
luna_foldernames = list(ID_df[ID_df['ID'].isin(luna_df['ID'])]['FolderName'])
```

```python
luna_foldernames
```

```python
study_df['Luna_features'] = [ 1 if row['FolderName'] in luna_foldernames else 0 for index,row in study_df.iterrows()]
```

```python
study_df
```

```python
coherence_df = pd.read_csv(r'..\eeg_data\coherence_df.csv')
```

```python
coherence_foldernames = coherence_df.dropna(subset=['Wake_O1-O2_Delta']).FolderName.unique()
```

```python
study_df['Coherence_features'] = [ 1 if row['FolderName'] in coherence_foldernames else 0 for index,row in study_df.iterrows()]
```

```python
print("Microfeatures - Available:", len(study_df[study_df['BA_Microfeatures'] == 1]),' Missing:', len(study_df[study_df['BA_Microfeatures'] == 0]),'')
print("Macrofeatures - Available:", len(study_df[study_df['Macrofeatures'] == 1]),' Missing:',len(study_df[study_df['Macrofeatures'] == 0]))
print("Alpha features - Available:", len(study_df[study_df['Alpha_features'] == 1]),' Missing:',len(study_df[study_df['Alpha_features'] == 0]))
print("Luna features - Available:", len(study_df[study_df['Luna_features'] == 1]), ' Missing:',len(study_df[study_df['Luna_features'] == 0]))
print("Coherence features - Available:", len(study_df[study_df['Coherence_features'] == 1]), ' Missing:',len(study_df[study_df['Coherence_features'] == 0]))
```
