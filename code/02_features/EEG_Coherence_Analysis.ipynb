```python
import pandas as pd
import import_ipynb
from cuzick_test import cuzick_test
import numpy as np
```

```python
#plot EEG coherence across all
```

```python
coherence_df = pd.read_csv('..\eeg_data\coherence_df.csv')
```

```python
coherence_df
```

```python
coherence_df = coherence_df.dropna(subset=['Wake_O1-O2_Theta'])
coherence_df = coherence_df.drop_duplicates()
```

```python
coherence_df 
```

```python
criteria_df = pd.read_csv('..\medical_data\study_criteria_table.csv')
criteria_df
```

```python
criteria_df.column
```

```python
dementia_studies = criteria_df[ (criteria_df['Dementia_Med'] ==1) | (criteria_df['Dementia_Prob'] == 1)]['FolderName']
```

```python
MCI_studies = criteria_df[ (criteria_df['MCI_Med'] ==1)  | (criteria_df['MCI_Prob'] == 1)]['FolderName']
```

```python
non_dementia_studies = criteria_df[ (criteria_df['Non_Enc'] ==1)]['FolderName']
```

```python
non_dementia_studies 
```

```python
coherence_stats_df = pd.DataFrame()
```

```python
coherence_stats_df['Dementia_Mean'] = coherence_df[coherence_df['FolderName'].isin(dementia_studies)].mean()
coherence_stats_df['Dementia_SE'] =  coherence_df[coherence_df['FolderName'].isin(dementia_studies)].sem()
```

```python
coherence_stats_df['MCI_Mean'] = coherence_df[coherence_df['FolderName'].isin(MCI_studies)].mean()
coherence_stats_df['MCI_SE'] = coherence_df[coherence_df['FolderName'].isin(MCI_studies)].sem()
```

```python
coherence_stats_df['NonDementia_Mean']  = coherence_df[coherence_df['FolderName'].isin(non_dementia_studies)].mean()
coherence_stats_df['NonDementia_SE']  = coherence_df[coherence_df['FolderName'].isin(non_dementia_studies)].sem()
```

```python
coherence_stats_df = coherence_stats_df.reset_index()
```

```python
coherence_stats_df.columns = ['Feature_Name', 'Dementia_Mean', 'Dementia_SE', 'MCI_Mean', 'MCI_SE',
       'NonDementia_Mean', 'NonDementia_SE']
```

```python
coherence_stats_df
```

```python
coherence_stats_df['TrendTest_p'] = np.nan
```

```python
for index,row in coherence_stats_df.iterrows():
    Feature = row['Feature_Name']
    t,p = cuzick_test([coherence_df[coherence_df['FolderName'].isin(dementia_studies)].dropna()[Feature],coherence_df[coherence_df['FolderName'].isin(MCI_studies)].dropna()[Feature],coherence_df[coherence_df['FolderName'].isin(non_dementia_studies)].dropna()[Feature]])
    coherence_stats_df.at[index,'TrendTest_p'] = p
    
```

```python
coherence_stats_df
```

```python
coherence_stats_df[coherence_stats_df['TrendTest_p'] < 0.05]
```

```python
coherence_stats_df[(coherence_stats_df['TrendTest_p'] < 0.05) & (coherence_stats_df['Dementia_Mean'] >coherence_stats_df['NonDementia_Mean'])  ]
```

```python
coherence_stats_df[(coherence_stats_df['TrendTest_p'] < 0.05) & (coherence_stats_df['Dementia_Mean'] >coherence_stats_df['NonDementia_Mean'])  ].sort_values('TrendTest_p')
```

```python
# Increase coherence with dementia
# C4-O2 
# F4-O2
# C3-O1
# F3-O1
# F3-O2
```

```python
coherence_stats_df[(coherence_stats_df['TrendTest_p'] < 0.05) & (coherence_stats_df['Dementia_Mean'] <coherence_stats_df['NonDementia_Mean'])  ].sort_values('TrendTest_p')[:50]
```
